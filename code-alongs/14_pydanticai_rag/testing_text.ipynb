```python
from pypdf import PdfReader
from backend.constants import DATA_PATH

path = DATA_PATH / "Dwarf_Hotot.pdf"
reader = PdfReader(path)
```

```python
all_text = ""
for page in reader.pages:
    text = page.extract_text()
    if text:
        all_text += text + "\n"
        
```
