## Deserialization - turn json object into Pydantic model

- data has a key called results
- inside of results it has value of a list of dictionaries
- another key called count
- choose the fields that you want to include through displaying data.keys()

```python
from pydantic import BaseModel # validate josn through pydantic BaseModel

class PokemonListResponse(BaseModel): #4 keys available. choose the ones preferred for app
    results: list[dict]
    count: int

pokemons = PokemonListResponse.model_validate(data)
pokemons
```

```python
pokemons.count
```

```python
pokemons.results # list of dictionaries
```

```python
from pydantic import computed_field

class PokemonListResponse(BaseModel):
    results: list[dict]
    count: int

# computed field
    @computed_field
#decorator:
    @property
    def number_stored_pokemons(self) -> int:
        return len(self.results)
    
pokemons = PokemonListResponse.model_validate(data)
pokemons.number_stored_pokemons #a field for counting number of data points received
```

## serialize into json
- pydantic model -> json

```python
with open("pokemons.json", "w") as file:
    file.write(pokemons.model_dump_json(indent=3)) #indent relevant for printing
```
