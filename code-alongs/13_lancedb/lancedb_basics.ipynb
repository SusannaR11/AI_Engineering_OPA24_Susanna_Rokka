## LanceDB -  a vectordatabase for LLM applications

```python
import lancedb

db = lancedb.connect(uri="vector_database")
db 
```

```python
db.uri
```

## Create a table

```python
import json
with open("animals_text_embedded.json", "r") as file:
    data= json.loads(file.read())

data
```

```python
table_animals = db.create_table("animals_text_embedded.json", exist_ok=True, data=data)
# add mode= "overwrite" for another version in vector_database
table_animals
```

```python
import pandas

table_animals.to_pandas()
```

## Add more data to vector database

```python
more_data = [
    {"text": "A panda eating bamboo peacefully.", "vector": [0.51, 0.37, 0.82]},
    {"text": "A lion roaring loudly on a rock.", "vector": [0.93, 0.18, 0.41]},
]
table_animals.add(more_data)
```

```python
table_animals.to_pandas()
```
