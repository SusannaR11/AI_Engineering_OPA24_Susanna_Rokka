```python
from pydantic import BaseModel, Field
```

```python
class Irisinput(BaseModel):
    sepal_length: float = Field(gt=4, lt=8.5)
    sepal_width: float = Field(gt=1.8, lt=5)
    petal_length: float = Field(gt=0.8, lt=7.5)
    petal_width: float = Field(gt=0, lt=3)

Irisinput
```

```python
# to check validation within class 'Irisinput'. any number outside of range will result in
# a validation error
Irisinput(sepal_length=4.2, sepal_width=2.5, petal_length=0.9, petal_width=1)
```

```python
payload = Irisinput(sepal_length=4.2, sepal_width=2.5, petal_length=0.9, petal_width=1)
payload
```

```python
#'model_dump()' returns a dictionary representation of the model
# this dict-representation model we can now make into Dataframe
payload.model_dump()
```

```python
import pandas as pd 

# model as Dataframe
pd.DataFrame(payload)
```

```python
pd.DataFrame(payload.model_dump(), index=[0]) 
```

### check if endpoint works using httpx

```python
import httpx

url = "http://127.0.0.1:8000/api/iris/v1/predict"

payload = {
  "sepal_length": 4.1,
  "sepal_width": 1.9,
  "petal_length": 6.9,
  "petal_width": 2.8
}
with httpx.Client(timeout=10) as client:
    response = client.post(url, json= payload)
    response.raise_for_status()
    
response
```
