```python
from dotenv import load_dotenv
import os , json
from google import genai
from google.genai import types
from pydantic import BaseModel

load_dotenv()

client= genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

with open("student_text_1.txt", encoding="utf-8") as f: s1 = f.read() 
with open("student_text_2.txt", encoding="utf-8") as f: s2 = f.read()

response=client.models.generate_content(
    model = "gemini-2.5-flash", 
    contents=[
        "Du är svenskalärare och ska betygsätta ett nationellt prov.", 
        "Betyget sätts enligt svensk grundskola (A-F)." "Bedöm tre dimensioner: aspekt, disposition, språk (1-3).", 
        
        "Betygsätt dessa filer:",
        f"Elev 1:\n{s1}",
        f"Elev 2:\n{s2}",
        "Returnera endast JSON",
],
config=types.GenerateContentConfig(response_mime_type="application/json"),
)

print(response.text)

# it is easier to add response_mime_type and response_schema as parameters for LLM
# response, gives more structured output
```
